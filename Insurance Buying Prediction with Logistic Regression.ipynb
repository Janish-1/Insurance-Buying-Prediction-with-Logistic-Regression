```python
# Logistic Regression
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression # Importing the logistic Regression
from sklearn.model_selection import train_test_split # Importing the Linear Model in case if error arises
from sklearn import metrics # To check accuracy
```

```python
df = pd.read_csv("Insurance_Data.csv")
df
```

```python
plt.scatter(df.age,df.bought_insurance)
plt.show()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df[['age']],df[['bought_insurance']],test_size = 0.2,random_state=5) 
# df is data inputed, 80% is the trained data and 20% is the testing data and random state is to fix the training data and testing data
# train_test_split is method for training our model
```

```python
reg = LogisticRegression()
```

```python
reg.fit(X_train,y_train)
# Fitting the data in 2-d Array as training data where X_train and y_train is 80 % of the data which is being trained to the
# Model
```

```python
yp = reg.predict(X_test)
yp
```

```python
y_test
```

```python
metrics.accuracy_score(yp,y_test)
# Accuracy of the metrics
```
